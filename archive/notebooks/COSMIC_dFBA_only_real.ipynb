```python
import torch, sys
sys.path.insert(0, 'nn')
from model import CosmicNNSurrogateEnhanced, RatePredictionHead

# Check rates are now time-varying
rph = RatePredictionHead(25, 64)
latent = torch.randn(4, 64)
time   = torch.rand(4, 13).sort(dim=1).values
g, p = rph(latent, time)

rate_std = g[0].std(dim=0).mean().item()
print(f"Rates std across time steps: {rate_std:.4f}")
print("GOOD - time-varying" if rate_std > 0.01 else "BAD - still constant")

# Check full model forward pass
model = CosmicNNSurrogateEnhanced(n_components=25, n_params=53, latent_dim=64)
ic     = torch.randn(4, 25)
params = torch.randn(4, 53)
out = model(ic, time, params)
print(f"concentrations shape: {out['concentrations'].shape}")  # expect (4, 13, 25)
```
